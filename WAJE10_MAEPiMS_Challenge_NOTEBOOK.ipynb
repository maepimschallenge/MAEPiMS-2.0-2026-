```python
data = pd.read_csv("nigeria_flu_weekly_national.csv")

train = data[data['season'] != '2025/2026']
test = data[data['season'] == '2025/2026'].copy()

weekly_average = train.groupby('epi_week_of_season')['cases'].mean()
test['forecast'] = test['epi_week_of_season'].map(weekly_average)

mae = (test['cases'] - test['forecast']).abs().mean()
print("Baseline MAE:", mae)
```

```python
season_2023 = train[train['season'] == '2023/2024'].set_index('epi_week_of_season')['cases']
season_2024 = train[train['season'] == '2024/2025'].set_index('epi_week_of_season')['cases']

print(season_2023.head(10))
print(season_2024.head(10))
```

```python
weighted_forecast = (season_2023 * 0.3) + (season_2024 * 0.7)
print(weighted_forecast.head(10))
```

```python
test['forecast_weighted'] = test['epi_week_of_season'].map(weighted_forecast)

mae_weighted = (test['cases'] - test['forecast_weighted']).abs().mean()
print("Weighted MAE:", mae_weighted)
print("Baseline MAE (for comparison):", mae)
```

```python


weekly_median = train.groupby('epi_week_of_season')['cases'].median()

test['forecast_median'] = test['epi_week_of_season'].map(weekly_median)

mae_median = (test['cases'] - test['forecast_median']).abs().mean()
print("Median-based MAE:", mae_median)
print("Baseline (average) MAE:", mae)
print("Weighted MAE:", mae_weighted)
```

```python

import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
def make_features(df):
    week = df['epi_week_of_season']
    features = pd.DataFrame({
        'week': week,
        'sin1': np.sin(2 * np.pi * week / 52),
        'cos1': np.cos(2 * np.pi * week / 52),
        'sin2': np.sin(4 * np.pi * week / 52),
        'cos2': np.cos(4 * np.pi * week / 52),
    })
    return features

X_train = make_features(train)
y_train = train['cases']

X_test = make_features(test)
y_test = test['cases']
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
test['forecast_harmonic'] = model.predict(X_test)

mae_harmonic = (test['cases'] - test['forecast_harmonic']).abs().mean()
print("Harmonic model MAE:", mae_harmonic)
print("Baseline MAE (for comparison):", mae)
```

```python


plt.figure(figsize=(12, 5))
plt.plot(test['epi_week_of_season'], test['cases'], label='Actual')
plt.plot(test['epi_week_of_season'], test['forecast_harmonic'], label='Harmonic forecast')
plt.plot(test['epi_week_of_season'], test['forecast'], label='Baseline forecast', linestyle='--')
plt.title('Actual vs Forecasts: 2025/2026 Season')
plt.xlabel('Week of season')
plt.ylabel('Cases')
plt.legend()
plt.show()
```

```python
print("Minimum forecast value:", test['forecast_harmonic'].min())
print("Maximum forecast value:", test['forecast_harmonic'].max())
print("Actual max:", test['cases'].max())
```

```python

def make_features_v2(df):
    week = df['epi_week_of_season']
    features = pd.DataFrame({'week': week})
    for k in [1, 2, 3, 4]:
        features[f'sin{k}'] = np.sin(2 * np.pi * k * week / 52)
        features[f'cos{k}'] = np.cos(2 * np.pi * k * week / 52)
    return features

X_train2 = make_features_v2(train)
X_test2 = make_features_v2(test)

model2 = LinearRegression()
model2.fit(X_train2, y_train)
test['forecast_harmonic2'] = model2.predict(X_test2)

mae_harmonic2 = (test['cases'] - test['forecast_harmonic2']).abs().mean()
print("Improved harmonic MAE:", mae_harmonic2)
```

```python
test['forecast_harmonic2_clipped'] = test['forecast_harmonic2'].clip(lower=0)

mae_harmonic2_clipped = (test['cases'] - test['forecast_harmonic2_clipped']).abs().mean()
print("Improved + clipped harmonic MAE:", mae_harmonic2_clipped)
```

```python
weekly_average_smooth = weekly_average.rolling(window=3, center=True, min_periods=1).mean()
print(weekly_average_smooth.head(10))
```

```python
test['forecast_smooth'] = test['epi_week_of_season'].map(weekly_average_smooth)

mae_smooth = (test['cases'] - test['forecast_smooth']).abs().mean()
print("Smoothed baseline MAE:", mae_smooth)
print("Original baseline MAE:", mae)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = pd.read_csv("nigeria_flu_weekly_national.csv")

train = data[data['season'] != '2025/2026']
test = data[data['season'] == '2025/2026'].copy()

weekly_average = train.groupby('epi_week_of_season')['cases'].mean()
test['forecast'] = test['epi_week_of_season'].map(weekly_average)

mae = (test['cases'] - test['forecast']).abs().mean()
print("Baseline MAE:", mae)
```

```python
weekly_std = train.groupby('epi_week_of_season')['cases'].std()
print(weekly_std.head(10))
```

```python
test['std'] = test['epi_week_of_season'].map(weekly_std)

# 90% interval (roughly, using a common statistical rule of thumb)
test['lower_90'] = (test['forecast'] - 1.645 * test['std']).clip(lower=0)
test['upper_90'] = test['forecast'] + 1.645 * test['std']

# 50% interval
test['lower_50'] = (test['forecast'] - 0.674 * test['std']).clip(lower=0)
test['upper_50'] = test['forecast'] + 0.674 * test['std']

test[['epi_week_of_season', 'cases', 'forecast', 'lower_50', 'upper_50', 'lower_90', 'upper_90']].head(10)
```

```python

inside_50 = ((test['cases'] >= test['lower_50']) & (test['cases'] <= test['upper_50'])).mean() * 100
inside_90 = ((test['cases'] >= test['lower_90']) & (test['cases'] <= test['upper_90'])).mean() * 100

print(f"% of weeks where actual falls inside 50% interval: {inside_50:.1f}%")
print(f"% of weeks where actual falls inside 90% interval: {inside_90:.1f}%")
```

```python
test['lower_50_v2'] = (test['forecast'] - 0.45 * test['std']).clip(lower=0)
test['upper_50_v2'] = test['forecast'] + 0.45 * test['std']

inside_50_v2 = ((test['cases'] >= test['lower_50_v2']) & (test['cases'] <= test['upper_50_v2'])).mean() * 100
print(f"Adjusted 50% interval catches: {inside_50_v2:.1f}%")
```

```python
weekly_avg_hosp = train.groupby('epi_week_of_season')['hospitalisation'].mean()
weekly_std_hosp = train.groupby('epi_week_of_season')['hospitalisation'].std()

test['forecast_hosp'] = test['epi_week_of_season'].map(weekly_avg_hosp)
test['std_hosp'] = test['epi_week_of_season'].map(weekly_std_hosp)

mae_hosp = (test['hospitalisation'] - test['forecast_hosp']).abs().mean()
print("Hospitalisation MAE:", mae_hosp)
```

```python
results = []

for state in state_data['state'].unique():
    state_subset = state_data[state_data['state'] == state]

    train_state = state_subset[state_subset['season'] != '2025/2026']
    test_state = state_subset[state_subset['season'] == '2025/2026'].copy()

    weekly_avg = train_state.groupby('epi_week_of_season')['cases'].mean()
    test_state['forecast'] = test_state['epi_week_of_season'].map(weekly_avg)

    mae_state = (test_state['cases'] - test_state['forecast']).abs().mean()
    pct_error = (mae_state / test_state['cases'].mean()) * 100

    results.append({'state': state, 'mae': mae_state, 'pct_error': pct_error})

results_df = pd.DataFrame(results)
results_df.sort_values('pct_error')
```

```python
def quantile_loss(actual, predicted, q):
    diff = actual - predicted
    return np.maximum(q * diff, (q - 1) * diff).mean()

losses = []
for q, col in zip([0.1, 0.25, 0.5, 0.75, 0.9], ['quantile_10', 'quantile_25', 'quantile_50', 'quantile_75', 'quantile_90']):
    loss = quantile_loss(test['cases'], test[col], q)
    losses.append(loss)
    print(f"Quantile loss at {q}:", loss)

quantile_loss_avg = np.mean(losses)
print("Average quantile loss:", quantile_loss_avg)
```

```python
def interval_score(actual, lower, upper, alpha):
    width = upper - lower
    penalty_low = (2 / alpha) * np.maximum(lower - actual, 0)
    penalty_high = (2 / alpha) * np.maximum(actual - upper, 0)
    return (width + penalty_low + penalty_high).mean()

wis_50 = interval_score(test['cases'], test['quantile_25'], test['quantile_75'], 0.5)
wis_90 = interval_score(test['cases'], test['quantile_10'], test['quantile_90'], 0.1)

wis = (0.5 * mae + 0.5 * (0.5 * wis_50) + 0.5 * (0.1 * wis_90)) / 2
print("WIS:", wis)
```

```python
taus = np.arange(0.01, 1.0, 0.01)

crps_values = []
for t in taus:
    q_forecast = stats.norm.ppf(t, loc=test['forecast'], scale=test['std']).clip(min=0)
    crps_values.append(quantile_loss(test['cases'], q_forecast, t))

crps = 2 * np.mean(crps_values)
print("CRPS:", crps)
```

```python
def check_coverage(actual, lower_50, upper_50, lower_90, upper_90):
    cov_50 = ((actual >= lower_50) & (actual <= upper_50)).mean() * 100
    cov_90 = ((actual >= lower_90) & (actual <= upper_90)).mean() * 100
    return cov_50, cov_90
```

```python
data = pd.read_csv("nigeria_flu_weekly_national.csv")

train = data[data['season'] != '2025/2026']
test = data[data['season'] == '2025/2026'].copy()

weekly_average = train.groupby('epi_week_of_season')['cases'].mean()
weekly_std = train.groupby('epi_week_of_season')['cases'].std()
test['forecast'] = test['epi_week_of_season'].map(weekly_average)
test['std'] = test['epi_week_of_season'].map(weekly_std)

for q in [0.1, 0.25, 0.5, 0.75, 0.9]:
    col_name = f'quantile_{int(q*100)}'
    test[col_name] = stats.norm.ppf(q, loc=test['forecast'], scale=test['std']).clip(min=0)

def check_coverage(actual, lower_50, upper_50, lower_90, upper_90):
    cov_50 = ((actual >= lower_50) & (actual <= upper_50)).mean() * 100
    cov_90 = ((actual >= lower_90) & (actual <= upper_90)).mean() * 100
    return cov_50, cov_90

state_data = pd.read_csv("nigeria_flu_weekly_by_state.csv")

def get_region(zone):
    if zone in ['NW', 'NE', 'NC']:
        return 'North'
    else:
        return 'South'

state_data['region'] = state_data['zone'].apply(get_region)

print("Ready — data shape:", data.shape, "| state_data shape:", state_data.shape)
```

```python
state_coverage = []

for state in state_data['state'].unique():
    state_subset = state_data[state_data['state'] == state]

    train_s = state_subset[state_subset['season'] != '2025/2026']
    test_s = state_subset[state_subset['season'] == '2025/2026'].copy()

    avg_s = train_s.groupby('epi_week_of_season')['cases'].mean()
    std_s = train_s.groupby('epi_week_of_season')['cases'].std()
    test_s['forecast'] = test_s['epi_week_of_season'].map(avg_s)
    test_s['std'] = test_s['epi_week_of_season'].map(std_s)

    q25 = stats.norm.ppf(0.25, loc=test_s['forecast'], scale=test_s['std']).clip(min=0)
    q75 = stats.norm.ppf(0.75, loc=test_s['forecast'], scale=test_s['std']).clip(min=0)
    q10 = stats.norm.ppf(0.10, loc=test_s['forecast'], scale=test_s['std']).clip(min=0)
    q90 = stats.norm.ppf(0.90, loc=test_s['forecast'], scale=test_s['std']).clip(min=0)

    cov_50, cov_90 = check_coverage(test_s['cases'], q25, q75, q10, q90)
    state_coverage.append({'state': state, 'cov_50': cov_50, 'cov_90': cov_90})

state_coverage_df = pd.DataFrame(state_coverage)
print("Average 50% coverage across all states:", state_coverage_df['cov_50'].mean())
print("Average 90% coverage across all states:", state_coverage_df['cov_90'].mean())
print(state_coverage_df.sort_values('cov_90').head(10))
```

```python
overall_relative_spread = (state_coverage_df['cov_50'].count())  # just a placeholder check
kappa = 0.7  # share of spread that scales with the state's own forecast size

state_coverage_v2 = []

for state in state_data['state'].unique():
    state_subset = state_data[state_data['state'] == state]
    train_s = state_subset[state_subset['season'] != '2025/2026']
    test_s = state_subset[state_subset['season'] == '2025/2026'].copy()

    avg_s = train_s.groupby('epi_week_of_season')['cases'].mean()
    test_s['forecast'] = test_s['epi_week_of_season'].map(avg_s)

    # spread now scales with the size of the forecast itself, not just that state's own 2-season std
    test_s['std_v2'] = kappa * test_s['forecast']

    q25 = stats.norm.ppf(0.25, loc=test_s['forecast'], scale=test_s['std_v2']).clip(min=0)
    q75 = stats.norm.ppf(0.75, loc=test_s['forecast'], scale=test_s['std_v2']).clip(min=0)
    q10 = stats.norm.ppf(0.10, loc=test_s['forecast'], scale=test_s['std_v2']).clip(min=0)
    q90 = stats.norm.ppf(0.90, loc=test_s['forecast'], scale=test_s['std_v2']).clip(min=0)

    cov_50, cov_90 = check_coverage(test_s['cases'], q25, q75, q10, q90)
    state_coverage_v2.append({'state': state, 'cov_50': cov_50, 'cov_90': cov_90})

state_coverage_v2_df = pd.DataFrame(state_coverage_v2)
print("NEW average 50% coverage:", state_coverage_v2_df['cov_50'].mean())
print("NEW average 90% coverage:", state_coverage_v2_df['cov_90'].mean())
```

```python



kappa = 0.7
b = 0.075  # a small floor based on that state's overall average

state_coverage_v3 = []

for state in state_data['state'].unique():
    state_subset = state_data[state_data['state'] == state]
    train_s = state_subset[state_subset['season'] != '2025/2026']
    test_s = state_subset[state_subset['season'] == '2025/2026'].copy()

    avg_s = train_s.groupby('epi_week_of_season')['cases'].mean()
    test_s['forecast'] = test_s['epi_week_of_season'].map(avg_s)

    overall_avg = train_s['cases'].mean()  # that state's typical case count, ignoring week
    test_s['std_v3'] = np.sqrt((kappa * test_s['forecast'])**2 + (b * overall_avg)**2)

    q25 = stats.norm.ppf(0.25, loc=test_s['forecast'], scale=test_s['std_v3']).clip(min=0)
    q75 = stats.norm.ppf(0.75, loc=test_s['forecast'], scale=test_s['std_v3']).clip(min=0)
    q10 = stats.norm.ppf(0.10, loc=test_s['forecast'], scale=test_s['std_v3']).clip(min=0)
    q90 = stats.norm.ppf(0.90, loc=test_s['forecast'], scale=test_s['std_v3']).clip(min=0)

    cov_50, cov_90 = check_coverage(test_s['cases'], q25, q75, q10, q90)
    state_coverage_v3.append({'state': state, 'cov_50': cov_50, 'cov_90': cov_90})

state_coverage_v3_df = pd.DataFrame(state_coverage_v3)
print("v3 average 50% coverage:", state_coverage_v3_df['cov_50'].mean())
print("v3 average 90% coverage:", state_coverage_v3_df['cov_90'].mean())
```

```python
def coverage_v3(subset_train, subset_test):
    avg = subset_train.groupby('epi_week_of_season')['cases'].mean()
    subset_test = subset_test.copy()
    subset_test['forecast'] = subset_test['epi_week_of_season'].map(avg)
    overall_avg = subset_train['cases'].mean()
    subset_test['std_v3'] = np.sqrt((kappa * subset_test['forecast'])**2 + (b * overall_avg)**2)
    q25 = stats.norm.ppf(0.25, loc=subset_test['forecast'], scale=subset_test['std_v3']).clip(min=0)
    q75 = stats.norm.ppf(0.75, loc=subset_test['forecast'], scale=subset_test['std_v3']).clip(min=0)
    q10 = stats.norm.ppf(0.10, loc=subset_test['forecast'], scale=subset_test['std_v3']).clip(min=0)
    q90 = stats.norm.ppf(0.90, loc=subset_test['forecast'], scale=subset_test['std_v3']).clip(min=0)
    return check_coverage(subset_test['cases'], q25, q75, q10, q90)

print("National:", coverage_v3(train, test))
```
